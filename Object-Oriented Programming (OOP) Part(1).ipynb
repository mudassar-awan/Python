📌 **What is OOP**?

In Python, everything is treated as an object, making OOP a natural and powerful approach.

## 🏗️ Constructor in Python (__init__)
📌 What is a Constructor?

A constructor is a special method that is automatically called when an object is created.

In Python, the constructor is defined using:
init()

 It is also called the initializer method.

## 🔑 Purpose of __init__
- Initialize object attributes
- Assign values when an object is created
- Set default or custom data

## **Important**
 - The self parameter is the reference to the current instance of the class , and it is used to acess variable that belongs to that class

```python
class Student:
  name="karan" # class Variable
  def __init__(self): # Constructor automatically invoke when the object is created
    print(self) # self is actually the object of the class it points to the object
    print("Adding new student in the data-base.. ")

s1 = Student() # constructor automatically invoke here
print(s1)
```

## 2. Parameterized Constructor

## 🧠 Class Attribute vs Instance Attribute in Python

## 🏷️ 1. Class Attribute
👉 **Definition**

A class attribute is shared by all objects (instances) of the class.

✅ **Key Points**
- Defined outside __init__
- Same value for every object
- Stored in the class itself

```python
class Student:

  schoolname = "Ajk Uni " # Class Attribute
  name = 'ananymous'      # class attribute

  def __init__ (self, name, marks): # Constructor
    self.name = name # object/instance Attribute
    self.marks = marks


s1 = Student("ajmal", 80)
print(f"{s1.name} got {s1.marks} from {s1.schoolname} ")

print(Student.schoolname) # we can also print the class Attribute in this waay
print(s1.name) # it always print object attribute becuase obj Attribute precedence > clas|s Attribute

print(Student.schoolname) # Printing the Class Attribute
```

## Methods in OOP (Python)

In Object-Oriented Programming, methods are functions defined inside a class that describe the behavior of objects.

👉 **In simple words**:
- A method = function that belongs to a class

## Types of Methods in Python
1. **Instance Methods**
2. **Class Methods**
3. **Static Methods**

## Instance Methods
- Work with object data
- Use self parameter
- Most commonly used

```python
# Instance Method
class Student:
  schoolname = "XYZ SCHOOL"
  def __init__(self,name,marks):
    self.name = name
    self.marks = marks

  # Instance/Object method always take self as a argument
  def displayinfo(self): # Instance Method
    print(f"Student name is {self.name} and he/she got {self.marks} marks")

  def get_marks(self): # Instance Method
    return self.marks

s1 =Student('kamil',90)
s1.displayinfo()
print(s1.get_marks())
```

## 2. 📘 Class Method
 🧠 Concept

A class method is a method that:

- Works with the class itself (not individual objects)
- Uses cls instead of self
- Defined using @classmethod

👉 It is mainly used to access or modify class variables

```python
class Student:
    school_name = "Aitchison School"   # class variable

    @classmethod
    def change_school(cls, name):
        cls.school_name = name


# Before change
print(Student.school_name)

# Change using class method
Student.change_school(" Lead School System ")

# After change
print(Student.school_name)

# Again changing the school name using the class method
Student.change_school("Ajk School")

# Important
#  Updating via class method overwrites the previous class variable value

print(Student.school_name)
```

## 3. 📘 Static Method in Python (OOP)
🧠 Concept

A static method is a method that:

- Belongs to a class
- Does NOT use self (object)
- Does NOT use cls (class)
- Works like a normal function, but is grouped inside a class

👉 It is defined using the decorator: @staticmethod

✅ In simple words:

"Put utility/helper functions inside a class for better organization"

```python
class Student:
  def __init__(self,name,marks):
    self.name = name
    self.marks = marks

  def display(self): # Instance/ Object Method
    print(f"{self.name} got marks in three different subject are {self.marks}")

  # The Staticmethod logic is related to the class, but It does not depend on object data or class data
  @staticmethod
  def greet():
    print("Welcome to the Student Portal")

s1 = Student("Ali",[22,88,99])
s1.display()
s1.greet()
```

 ## 1️⃣ Abstraction (Hiding Complexity)

**Abstraction** is about showing only the necessary details to the user, while hiding internal implementations.

In Python, we often implement abstract classes using the abc module.

```python
class Car:
    def __init__(self):
        # Private attributes hidden from the user
        #Private Attribute: A variable in a class that is hidden from outside access to protect the internal state.
        self.__accelerator = False
        self.__brake = False
        self.__clutch = False

        # Abstract Method: A method that hides implementation details and exposes only the necessary functionality to the user.
    def start(self):
        # Abstracted method: user just calls start() without worrying about details
        self.__clutch = True
        self.__accelerator = True
        print("Car - Started!")

# Using the class
c1 = Car()
c1.start()
```

## 2️⃣ Encapsulation (Data Hiding)

Encapsulation is about restricting direct access to some attributes and exposing methods to interact safely with them.
- It is basically wrapping data and functions into the single unit (object)

## Private Attributes:

Variables in a class prefixed with __ that cannot be accessed directly from outside of the class , used to hide data and enforce controlled access through methods.

## Getter Method

A getter is a method used to access the value of a private attribute from outside the class.

Since private attributes (like __balance) cannot be accessed directly, the getter provides safe read access.

## Setter Method

A setter is a method used to modify the value of a private attribute from outside the class.

It allows controlled write access, often with validation.

**Problem**
- Create Account clas having two Attributes , Balance & Account-no
- Create the Method for Debit, Credit , and printing the Balance

```python
class Account:
    def __init__(self, balance, account_no):
        # Private attributes (encapsulation)
        self.__balance = balance        # Balance is hidden from direct access
        self.__account_no = account_no  # Account number is hidden from direct access

    def debit(self, amount):
        # Instance method to withdraw money safely
        self.__balance -= amount
        print(f"{amount} debited. {self.get_balance()}")  # Use getter to display balance

    def credit(self, amount):
        # Instance method to deposit money safely
        self.__balance += amount
        print(f"{amount} credited. {self.get_balance()}")  # Use getter to display balance

    def get_balance(self):
        # Getter method to safely access the private balance
        return f"Balance is {self.__balance}"


# Create an object of Account
account1 = Account(10000, 12345)  # Encapsulation ensures balance and account_no are private
print(account1.get_balance())
# Call methods to modify/access balance
account1.debit(100)   # Withdraw 100
account1.credit(500)  # Deposit 500

# Safely access balance using getter
print(account1.get_balance())
```

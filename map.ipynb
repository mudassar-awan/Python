## 🔹 Introduction <a id="introduction"></a>

Python supports **functional programming**, a style where functions operate on other functions.

Three important built-in functions:
- ✅ `map()` → transforms data  
- ✅ `filter()` → selects data  
- ✅ `reduce()` → aggregates data

Let's explore each with examples.

## 1️⃣ Understanding `map()` <a id="understanding-map"></a>

The map() function applies a specific function to every item in an iterable and returns an iterator with the results.

**Syntax:** 
- map(function, iterable)

**Use Case  🎯:** 
- When you want to transform every element (e.g., squaring every number in a list).


```python
x=3
def square(x):
  return x*x # x**2
print(square(x))

alist=[1,2,3,4,5]
def square(x):
  return x*x
print(map(square,alist))
print(list(map(square,alist)))
```

### 📝 Explanation:
- `square()` returns square of a number
- `map()` applies `square()` to each list element
- Convert map object to list for display

## 2️⃣ Using `map()` with Lambda <a id="using-map-with-lambda"></a>

### 📖 Lambda Function:
- Anonymous A small, one-line function without a name.
A lambda function can take any number of arguments, but it can only have one expression. The result of that expression is automatically returned.

### Syntax
- The structure is always:
- **lambda arguments : expression**

```python
# by using lamda fucntion
alist=[1,2,3,4,5]
print(map(lambda x:x*x,alist))
print(list(map(lambda x:x*x,alist)))
```

## 3️⃣ Understanding `filter()` <a id="understanding-filter"></a>

The **`filter()`** function constructs an iterator from elements of an iterable for which a function returns True.

**Syntax:** 
- filter(function, iterable)

**Use Case:** 
- When you want to extract a subset of data based on a condition (e.g., picking only even numbers).\
**Think of it as:**
- ➡️ “Keep what is True.”

```python
# filter function use
alist=[1,2,3,4,5,6]
even =list(filter(lambda x:x%2==0,alist))
print(even)
```

## 4️⃣ Understanding `reduce()` <a id="understanding-reduce"></a>
Unlike the others, `reduce()` is not a built-in function in the global namespace; you must import it from the functools module. It applies a rolling computation to sequential pairs of values in a list until only a single value remains.

`Syntax:` reduce(function, iterable)

`Use Case:` When you want to "shrink" a list down to one result (e.g., calculating the product or sum of all elements).

```python
from functools import reduce
alist=[1,2,3,4,5,6]
sum=reduce(lambda total,y:total+y,alist)
print(f"total sum of the list is {sum}")
```

```python
from functools import reduce
alist=[1,2,3,4]
mul=reduce(lambda total,y:total*y,alist)
print(f"total Multiply  of the list is {mul}")
```

## ✅ Summary <a id="summary"></a>

| Function | Purpose | Example |
|--------|--------|--------|
| map | Transform | Square numbers |
| filter | Select | Even numbers |
| reduce | Combine | Sum or Product |

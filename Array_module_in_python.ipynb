## 📦 Arrays in Python (Using array Module)
🔹 **Introduction**

In Python, the **built-in list** is commonly used, but the array module provides a more memory-efficient way to store homogeneous data (elements of the same type).

The array module is useful when you need compact storage and faster numerical operations compared to lists.

```python
# Arrays in python
lst=[1,2,3,5,5,4,"ali",False] # list can store multiple types of the data therefore  it is Heterogenous
print(lst)
print(type(lst))
print(len(lst))
```

## Importing the Module

```python
## importing array .
import array as arr
arary=arr.array('f',[1,2.3,3,4,3,2]) # Array just store one type of the data therefore it is Homogenous
print(type(arary))
print(arary)
```

##  Importing all function of the array

```python
from array import * # it import Everything from array module
a=array('i',[1,2,3,4,5,6]) # 'i' is indicating the type of data(int) that is stored in the array
print(type(a))
print(a)
```

```python
from array import array
# These numbers have higher accuracy compared to normal floats
arr = array('d', [1.23456789, 2.98765432, 3.14159265]) #
# Create an array of type 'd' (double precision float)

# Print the entire array
print("Double array:", arr)

for value in arr:
    # Print each value one by one
    print(value)
```

```python
from array import array
# Create an array of type 'b' (signed integers)
# Range allowed: -128 to 127
arr = array('b', [-10, 0, 25, 100])

print("Signed char array:", arr) # Print the Entire element of the array

for num in arr:
    # Print each number
    print(num)
```

```python
from array import array
# Create an array of type 'u' (unicode characters)
# Each element must be a single character (supports multiple languages)
arr = array('u', ['A', 'ب', '3'])

# Print the entire array
print("Unicode array:", arr)

# Loop through each character in the array
for char in arr:
    # Print each character
    print(char)
```

## 1. append operation in array

```python
from array import *
a=array('f',[1,2,3,4,5,6]) # 'f' is the float data type which is used to store floating data
a.append(10) # append function add the valus at the last index of the array
print(a)
```

## 2. insert operation in array

```python
from array import *
a=array('i',[1,2,3,4,5,6])
a.insert(0,10) # inserting the value 10 at index 0
print(a)
print(a[0])
```

## 3.remove operation in the array

```python
from array import *
a=array('i',[2,3,4,5,4,32])
a.remove(32) # removing the value (32) from the array
print(a)
```

## 4. updating index value  in array

```python
from array import *
a=array('i',[1,2,3,4,5,6])
print(a[2])
d=a[2]=32 # updating the value of index 2 from 3 to 32
print(f"update value at index 2 is {d}")
print(a)
```

```python
from array import array
# Create an array of type 'u' (unicode characters)
# Each element must be a single character (supports multiple languages)
arr = array('u', ['A', 'ب', '3'])
arr[2] = 'B'
# Print the entire array
print("Unicode array:", arr)

# Loop through each character in the array
for char in arr:
    # Print each character
    print(char)
```

## 5.converting array to list


```python
from array import *
a=array('i',[1,2,3,4,5,6])
lst=a.tolist() # by using this method we can convert the array into list
print(lst)
for x in lst:
  if x==3:
    break # terminating the loop when condition meets
  print(x)
```

## 6.looping the array


```python
from array import *
a=array('i',[10,2,3,4,55,66])
for value in a:
  print(value)
```

## 7. copying the array


```python
from array import *
a=array('i',[10,23,44,55,66])
newarr=a # copying the array
print(newarr)
a[2]=60 # updating the index (2) value to 60
print(a)
print(newarr)
```

```python
from array import *
a=array('i',[10,23,44,55,66])
arrcopy=array('i',a.tolist())
print(arrcopy)
```

## 8. Comparsion table of Python list & Python Array Module

| Feature | Python List | Python `array` Module |
| :--- | :--- | :--- |
| **Data Type** | **Heterogeneous** (can store mixed types like `[1, "Hi"]`) | **Homogeneous** (must be the same type, e.g., all integers) |
| **Import Required** | No (Built-in) | **Yes** (`import array`) |
| **Type Constraints** | None (Flexible) | **Strict** (Must specify a type code like `'i'` for int, `'f'` for float) |
| **Memory Efficiency** | **Lower** (Each element is a full Python object) | **Higher** (Stores raw values, much more compact) |
| **Performance** | Faster for general-purpose tasks | Faster for large sequences of numbers/basic I/O |
| **Functionality** | Extremely rich (methods for all data types) | Limited (focused on numeric storage and basic methods) |
| **Usage Syntax** | `my_list = [1, 2, 3]` | `my_arr = array.array('i', [1, 2, 3])` |
| **Best Use Case** | General programming, mixed data, ease of use | Efficient storage of large numeric data without NumPy |

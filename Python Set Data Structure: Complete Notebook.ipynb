**Introduction  of Set**
- A Set is an unordered collection of distinct, hashable objects.
- It is a data structure that implements the mathematical concept of a finite set.

Unlike sequences (such as lists or tuples), a set does not maintain a specific position for its elements and strictly prohibits duplicate values.

A set is defined by curly braces {} or the set() constructor.

````python
# Creating sets
fruits = {"apple", "banana", "cherry", "apple"}
print(f"Set automatically removes duplicates: {fruits}")

# Creating an empty set (Note: {} creates a dictionary)
empty_set = set()
print(type(empty_set))
````

````python
s={2,3333,4,5,6,7,7,7,7,7} # support unique values ignore repetation
print(s)
print(type(s))
````

## Some Set_Methods

## 1. Set .update() Method
- set.update() method adds all unique elements from an iterable (like a list or another set) into the original set in-place.


````python
set1={1,2,3,4,5,6,7,8}
set2={11,22,33,44,55}
set1.update(set2) # it add the whole whole of set 2 in set 1
print(set1)
````

## 2. Set .add() Method
- set.add() method inserts a single, hashable element into a set; if the element already exists, the set remains unchanged.


````python
s={'a','b','c','d','e'}
s.add('f') # adding 'f' into set(s)
print(s)
````

````python
numbers={1,2,3,4,5,6,7}
numbers.add(8)
numbers.add(9) # adding 9 into set (numbers)
print(numbers)
````

## 3. Set .union() Method
- set.union() method (or | operator) returns a new set containing all unique elements from two or more sets, effectively merging them.


````python
row1 = {1,2,3,4,5}
row2 = {2,33,4,55,5,6}
row3 = row1.union(row2)
row4 = row1 | row2 # it can also be done in this way
print(row3)
print(row4)
````

## 4. Set .intersection() Method
-  set .intersection() method (or & operator) returns a new set containing only the elements that are present in both (or all) specified sets.


````python
number1 = {1,2,3,4,5,6}
number2 = {2,33,4,111,888}
number3=number1.intersection(number2)
number4=number1 &  number2 # it can also be done in this way
print(number3)
print(number4)
````

## 5. Set .clear() Method
- set.clear() method removes all elements from the set, leaving it completely empty without deleting the set object itself.

````python
active_users = {'ajmal','kamil','Malik','Faizan'}
print(active_users)
active_users.clear() # it clear all the object from the set
print(active_users)
````

## 6. Set .copy() Method
- set.copy() method creates a shallow copy of a set, allowing you to manipulate a new version without affecting the original data.

````python
origional = {'data1','data2','data3'}
copied_data = origional.copy()
copied_data.add('data4')
print(origional)
print(copied_data)
````

## 7. Set .remove() Method
- set.remove() method search and  deletes a specific element from the set and raises a KeyError if the element is not found.

````python
elements = {2,3,5,6,7,8,9}
elements.remove(6) # it remove the 6 from the elements
print(elements)
````

## 9. Set .discard() Method
- The .discard() method removes a specific element from the set if it exists, but unlike .remove(), it does not raise an error if the element is missing.

**Important**
- Use .discard() for "safe" deletion when you aren't 100% certain the element exists in the set.

````python
# A set of active user IDs
active_users = {"user_101", "user_102", "user_103"}

# Use discard to remove an ID
active_users.discard("user_101")  # Removes "user_101"
active_users.discard("user_999")  # "user_999" is not in the set, but NO error is raised

print(active_users)
# Output: {'user_102', 'user_103'}
````

## 9. Set .differece() Method
- set_a.difference(set_b): Returns a new set containing elements that exist in the first set but not in the second, representing the mathematical $A - B$ operation.

````python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Difference: What does 'a' have that 'b' doesn't?
print(set_a - set_b)           # Output: {1, 2}
print(set_a.difference(set_b)) # it can also be done in this way
````

## 10. Set .symmetric_difference() method
- set_a.symmetric_difference(set_b): Returns a new set containing elements that are in either set, but not in both, effectively excluding the intersection.

````python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Symmetric Difference: What is unique to each individual set?
set_c = set_a.symmetric_difference(set_b)
print(set_c)
print(set_a ^ set_b) # it can also done in this way           # Output: {1, 2, 4, 5}
````

## 11. Set . isdisjoint() method
 - set.isdisjoint() method returns a boolean value (True or False) indicating whether two sets have zero elements in common.
 - set_a.isdisjoint(set_b): Returns True if the intersection of the two sets is empty, meaning they share no members at all.

````python
group_a = {"User_1", "User_2"}
group_b = {"User_3", "User_4"}

# Check if anyone is in both groups by mistake
if group_a.isdisjoint(group_b):
    print("Clean data: No overlap found.")
````

````python
group_a = {"User_1", "User_2"}
group_b = {"User_3", "User_4", "User_1"}

# Use 'not' to check if they ARE overlapping
if not group_a.isdisjoint(group_b):
    print("Overlap found! Someone is in both groups.")
else:
    print("Clean data: No overlap found.")
````

## 12. Set.issubset() Method
- set_a.issubset(set_b): Returns True if all elements of the first set are contained within the second set, representing the mathematical $A \subseteq B$ relationship.

````python
#.issubset() (Is "A" inside "B"?)
#Use this to check if a specific collection is entirely contained within another.
# The "Must-Have" list
required_ingredients = {"flour", "sugar", "eggs"}

# What we actually have in the kitchen
my_pantry = {"flour", "sugar", "eggs", "milk", "butter", "salt"}

# Is the recipe possible? (Are all requirements inside the pantry?)
can_bake = required_ingredients.issubset(my_pantry)

print(can_bake) # Output: True
````

## 13. Set.issuperset() Method
- set_a.issuperset(set_b): Returns True if the first set contains every element of the second set (and potentially more), representing the mathematical $A \supseteq B$ relationship.
- In Python, you can use comparison operators as a shortcut:

Subset: set_a <= set_b

Superset: set_a >= set_b

````python
# A streaming service library
netflix_library = {"Stranger Things", "The Crown", "Inception", "Breaking Bad"}

# A user's watchlist
user_watchlist = {"Inception", "The Crown"}

# Does Netflix have everything the user wants to watch?
has_all_movies = netflix_library.issuperset(user_watchlist)

print(has_all_movies) # Output: True
````

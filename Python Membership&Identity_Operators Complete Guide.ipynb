## **Membership Operators**
Membership operators are used to check whether a value exists in a sequence (like list, tuple, string, dictionary, etc.).

 ## ✅ Types of Membership Operators
**Operator	Meaning**
- **in**	Returns True if value exists
- **not in**	Returns True if value does NOT exist

## **in  (Membership Test):**
 Returns True if the specified value exists in the given sequence.


## ✅ String Example (Membership in Operator)

```python
text="alpha"
x=('a' ) in text
print(x)
x=('A') in text # Python is case senstive language
print(x)
```

## ✅ Tuple Example (Membership in Operator)

---



---



```python
numbers = (10, 20, 30, 40)

print(20 in numbers)      # True
if 10  in numbers:
  print(f"10 is  in {numbers} ")
```

## ✅ List Example (Membership in Operator)

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)   # True
if "apple" in fruits:
  print(f"apple is in {fruits}")
```

## ✅ Dictionary Example (Membership in  Operator)
📌 **Important:**
- 👉 Membership operator checks keys only in dictionaries.
- it doesnot check values

```python
data = {"name": "Ali", "age": 20}

print("name" in data)   # True (checks keys)
print("Ali" in data)    # False (values not checked)

if 'age' in data:
  print(f'age is present in {data.keys()}' )
```

## **not in  (Negative Membership Test):**
 Returns True if the specified value does not exist in the given sequence.

---



---




## ✅ String Example (Membership **not in** Operator)

```python
text="alpha"
x=('b' ) not in text
print(x)
x=('B')  not in text
print(x)
```

## ✅ List Example (Membership **not in** Operator)

```python
fruits = ["apple", "banana", "mango"]

print("apple" not in fruits)   # False
if 'grapes' not in fruits:     # True
  print(f'grapes are not in {fruits}')

```

## ✅ Dictionary Example (Membership **not in** Operator)
📌 Important:

- 👉 Membership operator checks keys only in dictionaries.
- it doesnot check values

```python
data = {"name": "Ali", "age": 20}

print("namee" not  in data)   # True (checks keys)
print("Ali" in data)    # False (values not checked)

if 'roll-no' not in data:    # True
  print(f'roll-no is not present in {data.keys()}' )
```

## **Identity Operators:**
In  Python, identity operators are used to compare whether two variables refer to the same object in memory, not just whether their values are equal.

### 🔹 Identity Operators in Python

**There are two identity operators:**

**Operator**	**Meaning**
- **is**	Returns True if both variables point to the same object
- **is not**	Returns True if variables point to different objects



**is      (Identity Test):** Returns True if both variables point to the same object in memory.


## ✅ List Example (Membership **is** Operator)

```python
a = [1, 2, 3]
b = a

print(a is b)   # True (same object)
```

###  Even if values are the same, different objects return False:

```python
x = [1, 2, 3]
y = [1, 2, 3]

print(x is y)   # False (different objects)
```

```python
x = None

if x is None:
    print("x is None")
```

```python
a = [1, 2]
b = [1, 2]

print(a == b)  # True (values same)
print(a is b)  # False (different objects)
```

```python
x=[1,22,33]
y=x
p=x is y
print(p)
# other case
z=[3,4,5]
d=x is z
print(d)
```

## 🔹 Identity Operator Confusing Cases
- 👉 Python usually caches integers in range **-5 to 256,** so they may share memory.

```python
# Case 1: Small integers (cached by Python)
a = 10
b = 10

print(a is b)   # True ✅ (Python caches small integers)
```

```python
# Case 2: Float values
x = 10.5
y = 10.5
# 👉 Floats are not cached like small integers → same value but different objects.
print(x is y)   # ❌ Usually False
print(x == y)   # True ✅
```

```python
# Case 3: Large integers (not always cached)
a = 1000
b = 1000

print(a is b)   # ❌ May be False (different objects)
# 👉 Large integers are not guaranteed to be cached → different memory locations.
```

```python
# Case 4: Strings (interning behavior)
s1 = "hello"
s2 = "hello"

print(s1 is s2)   # True ✅ (string interning)
```

**is not  (Negative Identity Test):**
- Returns True if both variables do not point to the same object in memory.
- Checks if two variables reference different objects.



## 🔹  String  Example (Membership **is not** Operator)
**Explanation**:
- Python may store same strings in same memory (string interning)
- So both point to same object
- Hence → is not = False

```python
s1 = "hello"
s2 = "hello"

print(s1 is not s2)  # False (in many cases)
```

##🔹 . List  Example (Membership **is not** Operator)

```python
x = [1, 2, 3]
y = [1, 2, 3]
 # here x and y belongs to different memory adresses
print(x is not y)   # True
```

## 🔹 . Tuple Example (Membership **is not** Operator)
**Explanation**:
- Both tuples have same values
- But stored at different memory locations
- So → is not returns True

```python
t1 = (1, 2, 3)
t2 = (1, 2, 3)

print(t1 is not t2)  # True
```

## Comparison table among these Operators

| Feature                  | Membership Operators (`in`, `not in`)              | Identity Operators (`is`, `is not`)            |
|--------------------------|--------------------------------------------------|----------------------------------------------|
| Purpose                  | Check if a value exists in a sequence            | Check if two variables refer to same object   |
| Operators                | `in`, `not in`                                  | `is`, `is not`                               |
| Works with               | Lists, tuples, sets, strings, dictionaries       | Any Python object                            |
| Comparison Type          | Value presence                                  | Object identity (memory location)            |
| Returns                  | `True` or `False`                               | `True` or `False`                            |
| Example                  | `'a' in "apple"` → `True`                        | `x is y`                                     |
| Opposite Operator        | `not in`                                        | `is not`                                     |
| Use Case                 | Searching elements in collections                | Checking if two variables are same instance  |
| Performance              | Depends on collection size                      | Very fast (just compares memory reference)   |
| Common Mistake           | Using for equality check                        | Using instead of `==` for value comparison   |
| Example Code             | `if 3 in [1,2,3]: print("Yes")`                 | `if x is y: print("Same object")`            |

## 🧠 del Keyword in Python
🔹 **What is del?**

The del keyword in Python is used to delete objects, variables, or references from memory.

In Object-Oriented Programming (OOP), del is commonly used to:

- Delete object instances
- Delete attributes of objects
- Remove references to objects (which may trigger object destruction)

```python
class Student:
  def __init__(self,name):
    self.name = name

s1 = Student("Tim") # s1 is the object of Student(class)
print(s1)
print(s1.name)

del(s1.name) # it will delete the name attribute of the object
del(s1)  # it will delete the object(s1)

```

## 🔒 Private Attributes in Python

Python doesn’t have true private variables like some languages (e.g., Java or C++), but it uses naming conventions to simulate privacy.

1. Double underscore (__attribute)
- Indicates: “internal use” (Strong convention)
- Not fully accessible from outside of class

```python
class Account:
  def __init__(self, acc_no, acc_pass):
    self.acc_no = acc_no
    self.__acc_pass = acc_pass # here __acc_pass is the private attribute it cannot be accessed directly from outside the class

  def reset_pass(self,):
    print(self.__acc_pass)

acc1 = Account(1233,"abcde")
print(acc1.acc_no)
# print(acc1.__acc_pass) it will raise an error due to private attribute

print(acc1.reset_pass())
```

## 🔒 Private Method (Python) —
 **Definition**

A private method in Python is a method defined with a double underscore (__) prefix, which triggers name mangling to limit direct access from outside the class.

**Important**
- Private attributes and methods are meant to used only within the class  and they are not accessible from outside the class directly


```python
class Person:
  __name = "anonymous" # private class attribute

  def __hello(self): # Private method we can't acces it through object method call
    print("hello Perosn!")

  def welcome(self):
    self .__hello()

p1 = Person()
print(p1.welcome())
# print(p1.__hello()) # it will raise an error bcz this method is private and not accessible directly
```

## 🔁 Inheritance in OOP (Python)

Inheritance is a concept where one class (child/subclass) acquires properties and methods of another class (parent/superclass).



## Types of Inheritance!
 **1. Single Inheritance**

One child inherits from one parent.
- 👉 Simple parent → child relationship

**2. Multiple Inheritance**

One child inherits from multiple parents.
- 👉 Combines features from multiple classes

**3. Multilevel Inheritance**

A chain of inheritance (grandparent → parent → child)
- 👉 Inheritance in levels

**4. Hierarchical Inheritance**

Multiple children inherit from one parent.
- 👉 One parent → many children

**5. Hybrid Inheritance**

Combination of multiple types (e.g., multiple + multilevel).
- 👉 Complex real-world structures

🧠 **Key Points**

Inheritance helps in:

- ✅ Code reuse
- ✅ Better organization
- ✅ Easier maintenance

 ## 1. **What is Single Inheritance?**

Single inheritance is a type of class inheritance where one child class inherits from only one parent class.

It allows the child class to:

Access methods and attributes of the parent class.
Extend or modify parent behavior.

**Conceptually:**

**Parent Class  →  Child Class**

Example of Single Inheritance

```python
class Car:
  color = "Black"    # class attribute

#Static Method
#Performs a task without accessing class or instance data.
#Static method: no self or cls, just utility functionality.

  @staticmethod
  def start():
    print("Car started")

  @staticmethod
  def stop():
    print("Car stopped")

class ToyotaCar(Car): # here ToyotaCar(class) is inherited by the Car(class)
  def __init__(self, name):
    self.name = name         # instance/object Attribute

car1 = ToyotaCar("Fortuner") # car1 is the 1st object of ToyotaCar(class)
car2 = ToyotaCar("Camry")    # car2 is the 2nd object of ToyotaCar(class)

print(car1.name)
print(f"color of the car is {car1.color}")

print(car1.start())
print(car1.stop())

print(f"{car2.name} is of {car2.color} color")
```

### 🔹 What is a Static Method in Python (OOP)?

A static method in Python is a method that:

- Belongs to a class
- Does NOT depend on the instance (self)
- Does NOT access class variables (cls)
- Works like a regular function, but is logically grouped inside a class


## 2. What is Multilevel Inheritance?

**Definition:**

Multilevel inheritance is a type of inheritance where a class derives from another class, which itself is derived from another class. Essentially, the inheritance chain has multiple levels.

**Structure:**

Grandparent Class → Parent Class → Child Class

**Example of Multi level Inheritance**

```python
class Car: # Grandparent class

  @staticmethod
  def start():
    print("Car started")

  @staticmethod
  def stop():
    print("Car stopped")

# Parent class
class ToyotaCar(Car):      # here ToyotaCar(class) is inherited by the Car(class)
  def __init__(self, brand):
    self.brand = brand     # instance attribute

# child class
class Fortuner(ToyotaCar): # here fortuner class is inherited by the ToyotaCar(class)
  def __init__(self, type):
    self.type = type      # instance attribute

  def get_type(self):
     print(f"Fortuner is {self.type}")

car1 = Fortuner("Diesel")
print(car1.start())
print(car1.get_type())
print(car1.stop())
```

## 3. What is Multiple Inheritance?

**Definition**:

Multiple inheritance is a type of inheritance where a child class inherits from more than one parent class.


- The Child Class gets attributes and methods from both Parent1 and Parent2.
- Python supports multiple inheritance directly.

**Example of  Multiple Inheritance**

```python
class A: # Grandparent class
  varA = "Welcome to Class A "

class B: # parent class
  varB = "Welcome to Class B "

# child class
class C(A,B): # Here class(C) is inherited by the two classes (A,B)
  varC = "Welcome to Class C "


c1 = C()
print(c1.varA)
print(c1.varB)
print(c1.varC)

```

```python
# Parent Class 1: Study-related features
class Student:

    def study(self):
        # This method shows student studying
        print("Student is studying subjects.")


# Parent Class 2: Sports-related features
class Athlete:

    def play_sport(self):
        # This method shows student playing sports
        print("Student is playing a sport.")


# Child Class: Inheriting from BOTH Student and Athlete
class UniversityStudent(Student, Athlete):

    def introduce(self):
        # This method is specific to UniversityStudent
        print("I am a university student.")


# 🔹 Create object
student1 = UniversityStudent()

# 🔹 Call methods from BOTH parent classes
student1.introduce()     # From child class
student1.study()         # From Student class
student1.play_sport()    # From Athlete class
```

## 4. What is Hierarchical Inheritance?

Definition:
Hierarchical inheritance is when multiple child classes inherit from a single parent class.

One parent → Many children
Each child can have its own methods and attributes in addition to the parent’s.

Structure:

        Parent
       /   |   \
Child1  Child2  Child3

- All children can access the parent’s methods.
- Each child can also have its own methods.

 **Hierarchical Inheritance Example**

```python
# Base/Parent  class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")


# Derived class 1
class Student(Person):
    def __init__(self, name, age, student_id):
        # Call parent constructor
        super().__init__(name, age)
        self.student_id = student_id

    def study(self):
        print(f"{self.name} is studying.")


# Derived class 2
class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def teach(self):
        print(f"{self.name} teaches {self.subject}.")


# Derived class 3
class Staff(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)
        self.department = department

    def work(self):
        print(f"{self.name} works in {self.department} department.")


# ------------------- Usage -------------------

student = Student("Ali", 20, "S123")
teacher = Teacher("Dr. Khan", 45, "Computer Science")
staff = Staff("Ahmed", 35, "Administration")

student.display_info()
student.study()

teacher.display_info()
teacher.teach()

staff.display_info()
staff.work()
```

## 5. Hybrid Inheritance

In object-oriented programming (especially in Python), hybrid inheritance is a combination of two or more types of inheritance in a single program.

Think of it as mixing patterns like:

- Single inheritance
- Multiple inheritance
- Multilevel inheritance
- Hierarchical inheritance
- Hybrid Inheritance Example

```python
# Base class
class Person:
    def __init__(self, name, **kwargs): # Accept and pass kwargs
        self.name = name
        super().__init__(**kwargs) # Call next in MRO (object's init)

    def show_name(self):
        print(f"Name: {self.name}")


# First derived class
class Student(Person):
    def __init__(self, name, student_id, **kwargs): # Accept and pass kwargs
        self.student_id = student_id
        super().__init__(name=name, **kwargs) # Pass specific args as kwargs, and rest as kwargs

    def study(self):
        print(f"{self.name} is studying.")


# Second derived class
class Employee(Person):
    def __init__(self, name, employee_id, **kwargs): # Accept and pass kwargs
        self.employee_id = employee_id
        super().__init__(name=name, **kwargs) # Pass specific args as kwargs, and rest as kwargs

    def work(self):
        print(f"{self.name} is working.")


# Hybrid class (Multiple inheritance)
class TeachingAssistant(Student, Employee):
    def __init__(self, name, student_id, employee_id):
        # Call super().__init__ with all arguments, using keyword arguments for clarity
        # The MRO will ensure each parent's __init__ gets its relevant arguments
        super().__init__(name=name, student_id=student_id, employee_id=employee_id)

    def assist(self):
        print(f"{self.name} is assisting in teaching.")


# ------------------- Usage -------------------

ta = TeachingAssistant("Sara", "S456", "E789")

ta.show_name()     # from Person
ta.study()         # from Student
ta.work()          # from Employee
ta.assist()        # own method
```

## super() method
What is super() in OOP?

👉 super() is used to call methods (especially constructors) from the parent class.

💡**Why use it?**
- Avoid rewriting code
- Reuse parent logic
- Maintain clean inheritance structure

```python
class Car:
  def __init__(self,type):
    self.type = type

  @staticmethod
  def start():
    print("Car Started")

  @staticmethod
  def stop():
    print("Car Stopped")

class ToyotaCar(Car):
  def __init__(self,name, type):
    super().__init__(type) # super method used for calling parent/super class Attributes/methods
    self.name = name
    super().start()
    super().stop()


car1 = ToyotaCar("parius", "Electric")
print(car1.type)
```

## 🔹 What is a Property Method?

A property lets you:

Access a method like an attribute
Add logic (validation, computation, etc.) when getting, setting, or deleting a value

It is created using the built-in **@property decorator.**
- **We use @property decorator on any method in the class to use the method as a property**

```python
class Student:
  def __init__(self, physics, chemistry, biology):
    self.physics = physics
    self.chemistry = chemistry
    self.biology = biology


 # def get_percentage(self):
   #     self.percentage = str((self.physics + self.chemistry + self.biology)//3) + "%"
  @property
  def percentage(self):
    return str((self.physics + self.chemistry + self.biology)//3) + "%"


s1 = Student(90,90,98)
print(s1.percentage)

s1.physics = 99
print(s1.physics)
print(s1.percentage)
```

## 🔹 What is a dunder function?

A dunder function stands for “double underscore” function — methods in Python that start and end with __.

These are also called special methods or magic methods.

🔹 **Key idea**

Dunder functions allow operator overloading and customization of built-in behavior.

```python
# Although python have a builtin data-type for Complex number .
# here we are understanding the concept of operator overloading through complex number by creating class
class Complex:
  def __init__(self, real, imagionary):
    self.real = real
    self.imagionary = imagionary

  def shownumber(self):
    print(f"{self.real}i +  {self.imagionary}j ")

  def __add__(self, num2):
    newreal = self.real + num2.real
    newimagionary = self.imagionary + num2.imagionary
    return Complex(newreal, newimagionary)

  def __sub__(self, num2):
    newreal = self.real - num2.real
    newimagionary = self.imagionary - num2.imagionary
    return Complex(newreal, newimagionary)

num1 = Complex(1,3)
num1.shownumber()


num2 = Complex(4,6)
num2.shownumber()

#num3 = num1 + num2
#num3.shownumber()

num3 = num1 - num2
num3.shownumber()

num3 = num2 - num1
num3.shownumber()
```

## 2.  🔹 Method Overloading in Python

Method overloading means defining multiple methods in the same class with the same name but different parameters (number or type of arguments).

It allows a method to behave differently depending on the arguments passed.

**Important in Python:**

Python does not support true method overloading like Java or C++. If you define multiple methods with the same name, the last one overwrites the previous ones. But you can achieve similar behavior using default arguments or *args/**kwargs

## 🔹 Example Using Default Arguments

```python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(5))        # Output: 5 (only a)
print(calc.add(5, 10))    # Output: 15 (a + b)
print(calc.add(5, 10, 15)) # Output: 30 (a + b + c)
```

Here, the add method works with 1, 2, or 3 arguments, mimicking method overloading.

## 🔹 Example Using *args

```python
class Calculator:
    def multiply(self, *args):
        result = 1
        for num in args:
            result *= num
        return result

calc = Calculator()
print(calc.multiply(2, 3))       # Output: 6
print(calc.multiply(2, 3, 4))    # Output: 24
```

Here, *args allows the method to handle any number of arguments.

## 🔹 Example Using *args & **kwargs

```python
def show_info(*args, **kwargs):
    print("Positional arguments:", list(args)) # list used here for positional arguments
    print("Keyword arguments:", kwargs)  # dictionary used for keyword arguments

# Calling the function
show_info(1, 2, 3, name="Alice", age=25)
```

✅ Explanation:

*args collects all positional arguments (1, 2, 3)

**kwargs collects all keyword arguments {'name': 'Alice', 'age': 25}

```python
class Circle:
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return (22/7) * self.radius * self.radius

  def perimeter(self):
    return 2 * (22/7) * self.radius

c1 = Circle(21)
print(c1.area())
print(c1.perimeter())
```

## Problem 1
Create a Employe class with attribute role , department , salary . This class also have show detail method

Create an Engineer class that inherits all attribute and method from Employe Class and this class should also contain two additional attributes: name and Age

```python
class Employe:
  def __init__ (self, role, department, salary):
    self.role = role
    self.department = department
    self.salary = salary

  def show_info(self):
    print(f"Role of the Employe is {self.role}")
    print(f"Department of the Employe is {self.department}")
    print(f"Salary of the Employe is {self.salary}")

class Engineer(Employe):
  def __init__(self,name,age):
    self.name = name
    self.age = age
    super().__init__("Engineer", "IT", "85,000")

  def show_info(self):
    super().show_info()
    print(f"Name of the Engineer is {self.name}")
    print(f"Age of the Engineer is {self.age}")
    return show_info
enng1 = Engineer("Jawad", "25")
enng1.show_info()
```

##  While loop
A while loop repeatedly executes a block of code as long as a condition is true.

✅ **Basic Syntax**

while condition:

    # code to run

🧠** How it works**

Python checks the condition.

If True → runs the loop body.

Repeats until the condition becomes False.

```python
i=1
while i<=4:
  print(i)
  i=i+1
```

⚠️ **Important: Infinite Loop**

If you don’t update the condition, the loop may run forever:

```python
"""while True:
    print("Hello")"""
```

## 5. Reverse a number

![Reversing a number in Python.png]()Reversing a number in Python.png

```python
# Take input from user
num = int(input("Enter a number: "))

# Initialize reversed number as 0
rev = 0

# Loop runs until num becomes 0
while num > 0:

    # Step 1: Extract last digit of number
    digit = num % 10

    # Step 2: Add digit to reversed number
    # First shift previous digits left (rev * 10)
    # then add new digit
    rev = rev * 10 + digit

    # Step 3: Remove last digit from original number
    num = num // 10

# Final output
print("Reversed number:", rev)

'''
print(end = "\n ")
n=input("Enter a number: ") # here we are taking the number as a string
revnum=n[::-1] this block of code reversing the string of number through slicing
print(f"reversed num are {revnum}")'''
```

## 8. Write a program that checks whether a given integer is a palindrome or not.

```python
# Take input from user
num = int(input("Enter a number: "))

# Initialize reversed number
rev = 0

# Store original number in a temporary variable
temp = num

# Loop to reverse the number
while temp > 0:
    digit = temp % 10        # Extract the last digit
    rev = rev * 10 + digit   # Append digit to reversed number
    temp = temp // 10        # Remove the last digit

# Check if the original number and reversed number are equal
if rev == num:
    print("Palindrome")
else:
    print("Not Palindrome")
```

## 9. Write a program that checks whether a given string  is a palindrome or not.

```python
# Simple palindrome check for a single word
word = input("Enter a word: ")

start = 0                      # points to the first character
end = len(word) - 1            # points to the last character

# Assume the word is a palindrome
is_palindrome = True

while start < end:

    # Compare characters from both ends
    if word[start] != word[end]:
        is_palindrome = False  # if mismatch, it's not a palindrome
        break                  # exit the loop early

    # Move pointers towards the center
    start += 1
    end -= 1

if is_palindrome:
    print("Word is Palindrome")
else:
    print("Word is not Palindrome")
```

## 13. Write a program that calculates the factorial of a  non-negative integer n.

```python
n = int(input("Enter a number "))
factorialn = 1
i = 1
while i <= n:
    factorialn *= i  # multiply factorialn by i
    i += 1
print(f"The factorial of {n} is {factorialn}")
```

## 16. Write a program to calculate the Sum of Digits

```python
n = int(input("Enter a number: "))
num = n  # keep original number if needed
sum_digits = 0

while n > 0:
    digit = n % 10       # get the last digit
    sum_digits += digit  # add it to sum
    n = n // 10          # remove the last digit

print(f"Sum of digits of {num} is {sum_digits}")
```
